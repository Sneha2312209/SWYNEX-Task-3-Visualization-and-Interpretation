```python
# SWYNEX Internship - Task 3: Visualization and Interpretation
# Downloads the SARS-CoV-2 reference genome from NCBI, analyzes it,
# creates three charts, and saves the results.

from Bio import Entrez, SeqIO
from collections import Counter
import matplotlib.pyplot as plt
import pandas as pd

# Replace with your own email address
Entrez.email = "your_email@example.com"

# 1. Download reference genome from NCBI
with Entrez.efetch(
    db="nuccore",
    id="NC_045512.2",
    rettype="fasta",
    retmode="text"
) as handle:
    record = SeqIO.read(handle, "fasta")

sequence = str(record.seq).upper()
valid_bases = ["A", "T", "G", "C"]
counts = Counter(sequence)

total_length = len(sequence)
base_counts = [counts.get(base, 0) for base in valid_bases]
ambiguous_count = total_length - sum(base_counts)

gc_content = (counts.get("G", 0) + counts.get("C", 0)) / total_length * 100
at_content = (counts.get("A", 0) + counts.get("T", 0)) / total_length * 100
frequencies = [count / total_length * 100 for count in base_counts]

print("Dataset:", record.id)
print("Genome length:", total_length)
print("A, T, G, C counts:", dict(zip(valid_bases, base_counts)))
print("GC content:", round(gc_content, 2), "%")
print("AT content:", round(at_content, 2), "%")
print("Ambiguous bases:", ambiguous_count)

# 2. Chart 1: nucleotide composition
plt.figure(figsize=(8, 5))
plt.bar(valid_bases, base_counts)
plt.title("SARS-CoV-2 Genome: Nucleotide Composition")
plt.xlabel("Nucleotide")
plt.ylabel("Number of bases")
plt.tight_layout()
plt.savefig("Chart1_Nucleotide_Composition.png", dpi=300)
plt.show()

# 3. Chart 2: GC and AT content
plt.figure(figsize=(7, 5))
plt.pie(
    [gc_content, at_content],
    labels=["GC content", "AT content"],
    autopct="%1.2f%%",
    startangle=90
)
plt.title("SARS-CoV-2 Genome: GC and AT Content")
plt.tight_layout()
plt.savefig("Chart2_GC_AT_Content.png", dpi=300)
plt.show()

# 4. Chart 3: nucleotide frequency
plt.figure(figsize=(8, 5))
plt.bar(valid_bases, frequencies)
plt.title("SARS-CoV-2 Genome: Nucleotide Frequency")
plt.xlabel("Nucleotide")
plt.ylabel("Frequency (%)")
plt.tight_layout()
plt.savefig("Chart3_Nucleotide_Frequency.png", dpi=300)
plt.show()

# 5. Save tabular results and FASTA sequence
results = pd.DataFrame({
    "Nucleotide": valid_bases,
    "Count": base_counts,
    "Frequency_Percentage": frequencies
})
results.to_csv("Nucleotide_Analysis.csv", index=False)

SeqIO.write(record, "Cleaned_SARS_CoV_2.fasta", "fasta")

print("\nFinished. The three PNG charts, CSV results, and FASTA file have been saved.")
```
